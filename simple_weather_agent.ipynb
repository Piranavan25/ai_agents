```python
# ✅ Full working agent setup with Groq (latest model) + Tavily search
# Make sure you have langchain, langchain_groq, and langgraph installed.

from langchain_groq import ChatGroq
from langchain_community.tools.tavily_search import TavilySearchResults
from langgraph.prebuilt import create_react_agent
from google.colab import userdata  # For securely fetching secrets in Colab

# --------------------------
# Step 1: Load API Keys
# --------------------------
GROQ_API_KEY = userdata.get('groq_API_key')
TAVILY_API_KEY = userdata.get('avily_API_key')

if not GROQ_API_KEY:
    raise ValueError("❌ GROQ_API_KEY not found. Please set it in Colab secrets.")
if not TAVILY_API_KEY:
    raise ValueError("❌ TAVILY_API_KEY not found. Please set it in Colab secrets.")

# --------------------------
# Step 2: Initialize the Groq LLM (latest recommended model)
# --------------------------
llm = ChatGroq(
    model="llama-3.3-70b-versatile",  # ✅ latest supported model
    temperature=0,
    api_key=GROQ_API_KEY
)

# --------------------------
# Step 3: Define Tools
# --------------------------
tools = [
    TavilySearchResults(max_results=2, tavily_api_key=TAVILY_API_KEY)
]

# --------------------------
# Step 4: Create the Agent
# --------------------------
agent = create_react_agent(llm, tools)

# --------------------------
# Step 5: Invoke the Agent
# --------------------------
inputs = {"messages": [("user", "What is the weather in San Francisco?")]}
result = agent.invoke(inputs)

print("\n🟢 Agent Response:")
print(result["messages"][-1].content)
```
